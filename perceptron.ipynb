# perceptron
$$\^{y}=f\left(\sum _{i=1}^{n}w_{i}x_{i}+b\right)$$
- $x_i$: The input features (the data you're feeding in).
- $w_i$: The weights, which determine the strength or "importance" of each input.
- $b$: The bias, which shifts the decision boundary to help the model fit the data better.
- $f$: The activation function (typically a Heaviside step function).

## Heaviside step function
$$H(x) = \begin{cases} 0, & x < 0 \\ 1, & x \geq 0 \end{cases}$$

(Note: In some mathematical contexts, $H(0)$ is defined as $0.5$, but in Computer Science and Neural Networks, it is almost always treated as $1$.)

The Modern Alternative: Sigmoid
While the Heaviside function was the foundation of early AI, modern neural networks usually use the Sigmoid or ReLU functions.

The problem with the Heaviside function is its "harshness." Because the jump is instantaneous, the slope (derivative) is either zero or undefined. This makes it impossible to use Gradient Descent (the math used to fine-tune modern AI), because you can't tell "how close" you were to the threshold; you're either in or you're out.
