# ReLU (Rectified Linear Unit)
## The "Workhorse" of Hidden Layers

ReLU is the most popular activation function for the internal (hidden) layers of a neural network. It is computationally efficient and helps prevent the model from "stalling" during training.
- Formula: $f(x) = \max(0, x)$
- Behavior: If the input is negative, it outputs 0. If the input is positive, it outputs the input exactly as it is.
- Best for: Hidden layers. It creates "sparse" networks where only the most important neurons stay active.

workhorse :
 a person who does most of the work in a group, or a machine/vehicle that is durable, dependable, and capable of heavy, sustained use

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate data
# We create 100 points between -10 and 10
x = np.linspace(-10, 10, 100)

# 2. Define the ReLU function
# np.maximum compares each element in x against 0
y = np.maximum(0, x)

# 3. Create the plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='ReLU', color='blue', linewidth=2)

# Add styling
plt.title('ReLU Activation Function', fontsize=14)
plt.xlabel('Input (x)')
plt.ylabel('Output (f(x))')
plt.grid(True, linestyle='--', alpha=0.7)
plt.axhline(0, color='black', lw=1) # Add x-axis line
plt.axvline(0, color='black', lw=1) # Add y-axis line
plt.legend()

# Display the plot
plt.show()
```

# Sigmoid
## The Binary Classifier
Sigmoid squashes any input value into a range between 0 and 1. Because it produces a smooth curve, it is perfect for models that need to predict a probability for a "Yes/No" question.
- Formula: $\sigma(x) = \frac{1}{1 + e^{-x}}$
- Behavior: Large positive numbers become almost 1; large negative numbers become almost 0.
- Best for: The output layer of a binary classification model (e.g., "Is this email spam or not?").

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 2. Generate a range of values from -10 to 10
# We use 100 points to make the curve look smooth
x = np.linspace(-10, 10, 100)
y = sigmoid(x)

# 3. Create the plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='Sigmoid Function', color='blue', linewidth=2)

# 4. Add formatting and labels
plt.title("Sigmoid Activation Function", fontsize=14)
plt.xlabel("Input (x)")
plt.ylabel("Output (y)")
plt.axhline(0, color='black', linewidth=0.5)  # x-axis
plt.axvline(0, color='black', linewidth=0.5)  # y-axis
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# Display the plot
plt.show()
```

# Softmax
## The Multi-Class Specialist

While Sigmoid is for "Yes/No," Softmax is for "Which one of these many options?" It takes a vector of scores and turns them into a probability distribution that sums up to exactly 1.0 (100%).

- Formula: $\sigma (\vec{z})_{i}=\frac{e^{z_{i}}}{\sum _{j=1}^{K}e^{z_{j}}}$

- Behavior: It highlights the largest value (making it "softly" the maximum) and suppresses the smaller values, ensuring they all add up to 1.

- Best for: The output layer of multi-class classification (e.g., "Is this a dog, a cat, or a bird?").

```python
import numpy as np
import matplotlib.pyplot as plt

def softmax(z):
    """Compute softmax values for each sets of scores in z."""
    e_z = np.exp(z - np.max(z)) # Subtracting max(z) for numerical stability
    return e_z / e_z.sum()

# 1. Prepare data
# We'll create a range of values to see how Softmax reacts to different inputs
x = np.linspace(-10, 10, 100)
# For visualization, let's assume we have a 3-class problem
# We vary one input (x) while keeping others constant
y1 = np.array([softmax([val, 2, 5]) for val in x])

# 2. Plotting
plt.figure(figsize=(10, 6))
plt.plot(x, y1[:, 0], label='Class 1 (Variable Input)', color='#1f77b4', lw=2)
plt.plot(x, y1[:, 1], label='Class 2 (Fixed Input = 2)', color='#ff7f0e', linestyle='--')
plt.plot(x, y1[:, 2], label='Class 3 (Fixed Input = 5)', color='#2ca02c', linestyle='--')

# Formatting the chart
plt.title('Softmax Function Behavior', fontsize=14)
plt.xlabel(r'Input Value ($z_i$)', fontsize=12)
plt.ylabel(r'Output Probability $\sigma(z)_i$', fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.ylim(-0.05, 1.05)

plt.show()
```
